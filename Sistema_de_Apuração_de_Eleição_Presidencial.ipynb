```python
"""
================================================================================
ATIVIDADE: Avaliação ACQA - Sistema de Apuração de Eleição Presidencial
DISCIPLINA: LABORATÓRIO DE PROGRAMAÇÃO / LINGUAGEM USADA: PYTHON
NOME: Daiane Aparecida Gomes Silva
DATA: 20/09/2026
================================================================================
"""

def main():

    # Canditados à presidência

    votos_lupa = 0
    votos_ciado = 0
    votos_bolsomilhas = 0
    votos_zebra = 0
    votos_nulos = 0
    votos_brancos = 0
    total_votos = 0

    # Telinha da urna

    print("=" * 45)
    print("BEM VINDO AO SISTEMA DE ELEIÇÃO PRESIDENCIAL 2026")
    print("\nPrepare sua colinha e escolha seu candidato")
    print(" \nVote com responsabilidade e sabedoria! Boa votação.")
    print("=" * 45)
    print("  Código | Candidato / Opção")
    print("  -------+-----------------------------------")
    print("     1   | Lupa")
    print("     2   | Ciado")
    print("     3   | Bolsomilhas")
    print("     4   | Zebra")
    print("     5   | Voto Nulo")
    print("     6   | Voto em Branco")
    print("     0   | Encerrar Votação")
    print("=" * 45)

    # Laço de confirmação e confirmação

# Laço de confirmação e votação
    while True:
        try:
            voto = int(input("\nDigite o número do seu voto (0 a 6): "))

            # Adicionei um botão de confirmação de encerramento para evitar encerrar "sem querer"

            if voto == 0:
                confirma = input("\nVocê realmente deseja encerrar a votação? (S/N): ").strip().upper()
                if confirma == "S":
                    print("\nEncerrando a votação e iniciando a apuração...")
                    break
                else:
                    print(">> Operação cancelada! A votação continua aberta.")

            # Contagem de votos

            elif voto == 1:
                votos_lupa += 1
                total_votos += 1
                print(">> [Sucesso] Voto confirmado para o candidato Lupa!")
            elif voto == 2:
                votos_ciado += 1
                total_votos += 1
                print(">> [Sucesso] Voto confirmado para o candidato Ciado!")
            elif voto == 3:
                votos_bolsomilhas += 1
                total_votos += 1
                print(">> [Sucesso] Voto confirmado para a candidata Bolsomilhas!")
            elif voto == 4:
                votos_zebra += 1
                total_votos += 1
                print(">> [Sucesso] Voto confirmado para a candidata Zebra!")
            elif voto == 5:
                votos_nulos += 1
                total_votos += 1
                print(">> [Atenção] Voto NULO computado com sucesso.")
            elif voto == 6:
                votos_brancos += 1
                total_votos += 1
                print(">> [Atenção] Voto em BRANCO computado com sucesso.")
            else:
                print(">> [Atenção] Número inválido! Digite apenas valores de 0 a 6.")

        except ValueError:
            print(">> Entrada inválida! Digite apenas números inteiros.")


    # Apuração dos resultados

    print("\n" + "=" * 45)
    print("            BOLETIM DE URNA / APURAÇÃO")
    print("=" * 45)

    # Se a urna não tiver nenhum voto

    if total_votos == 0:
        print("Nenhum voto foi registrado nesta eleição.")
        print("=" * 45)
        return

    # Percentual dos votos branco e nulos

    percentual_nulos = (votos_nulos / total_votos) * 100
    percentual_brancos = (votos_brancos / total_votos) * 100

    # Valores totais calculados

    print(f"Total de votos - 1 (Lupa):  {votos_lupa}")
    print(f"Total de votos - 2 (Ciado):  {votos_ciado}")
    print(f"Total de votos - 3 (Bolsomilhas): {votos_bolsomilhas}")
    print(f"Total de votos - 4 (Zebra):   {votos_zebra}")
    print("-" * 45)
    print(f"Total de votos nulos:       {votos_nulos}")
    print(f"Total de votos em branco:   {votos_brancos}")
    print(f"Total geral de votos:       {total_votos}")
    print("-" * 45)
    print(f"Percentual de votos nulos:  {percentual_nulos:.2f}%")
    print(f"Percentual de votos brancos:{percentual_brancos:.2f}%")
    print("=" * 45)

    # Canditado vencedor

    maior_voto = max(votos_lupa, votos_ciado, votos_bolsomilhas, votos_zebra)

    if maior_voto == 0:
        print("Resultado: Nenhum candidato recebeu votos válidos.")
    else:
        vencedores = []
        if votos_lupa == maior_voto:
            vencedores.append("Lupa")
        if votos_ciado == maior_voto:
            vencedores.append("Ciado")
        if votos_bolsomilhas == maior_voto:
            vencedores.append("Bolsomilhas")
        if votos_zebra == maior_voto:
            vencedores.append("Zebra")

        if len(vencedores) == 1:
            print(f"O candidato vencedor foi: {vencedores[0]} com {maior_voto} voto(s). Parabéns!")
        else:
            print(f"Empate entre os candidatos: {', '.join(vencedores)} com {maior_voto} voto(s) cada. Votação segue para segundo turno")

    print("=" * 45)


if __name__ == "__main__":
    main()


```
